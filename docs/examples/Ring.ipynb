```javascript
const { GraphElement } = await import("https://unpkg.com/@verticesandedges/graph-element/dist/index.js");

ui = document.createElement('div');

graph = document.createElement('graph-el');
graph.style.height = "500px";
ui.appendChild(graph);

class Cycle {
  constructor(){
    this.vertices = [];
    this.edges = [];
  }
  
  add(){
    this.eraseRing();
    const vertex = graph.addVertex();
    this.vertices.unshift(vertex);
    this.drawRing();
  }

  remove(){
    this.eraseRing();
    const vertex = this.vertices.shift();
    if(!vertex) return;
    vertex.remove();
    this.drawRing();
  }

  drawRing(){
    let first = false;
    let previous = false;
    this.vertices.forEach(current => {
      if(!first){
        first = current;
      }

      if(previous){
        this.edges.push(graph.addEdge(previous, current));
      }

      previous = current;
    });

    if(first === previous) return;
    this.edges.push(graph.addEdge(previous, first, { spline: previous === first ? true : false } ));
  }

  eraseRing(){
    let edge;
    while(edge = this.edges.shift()){
      edge.remove();
    }
  }
}

const cycle = new Cycle();

const addBtn = document.createElement('button');
addBtn.textContent = "Add Vertex";
addBtn.addEventListener("click", () => cycle.add() );
ui.appendChild(addBtn);

const removeBtn = document.createElement('button');
removeBtn.textContent = "Remove Vertex";
removeBtn.addEventListener("click", () => cycle.remove() );
ui.appendChild(removeBtn);

return ui;
```
