# Javascript Notebook Manual

## Purpose
The original point of the Javascript Notebook was as a frontend for a self modifying server. The server would have had an api for adding routes and frontend via request. A few years later, I was able to sit down and flesh out the concept. I wrote a AngularIO version with the spiral to the side, rewrote it in web components, polished it a bit, and published it to the Google Play and Microsoft Stores. 

This document and the examples at the end should serve to familiarize you with this client side javascript/typescript notebook.

## Usage
The notebook consists of markdown cells, and typescript/javascript cells. Markdown cells are for your prose and can be rendered (turned from plain text into pretty html), typescript and javascript cells hold code, and can be run (the contents of the cell are wrapped in an AsyncFunction, so await and return are available). Code cells can display numbers, booleans, strings, lists, objects, and HTMLElements.

To render or run a cell, you can either click the corresponding button below, or click inside the text field and press Ctrl-Enter (if you’re on a desktop or laptop). To continue writing in a markdown cell that’s already been rendered, simply double click/tap it, and the cell reverts to its editable form. 

Code cells can call console.log and console.error (other console functions will return an undefined-error). Markdown cells can render math symbols via LaTeX: $...$ (single dollar signs as delimiters) for inline rendering, and $$...$$ (double dollar signs as delimiters) for math blocks. Code cells can use global variables to make one variable available to subsequently run cells. Recall that a global variable gets declared without var, const or let.

## Context and Customization
The notebook’s code cells can be customized by setting a variable in the notebook’s context property. This property is a simple Javascript object. A name of the context’s field becomes a variable in a code cell.
