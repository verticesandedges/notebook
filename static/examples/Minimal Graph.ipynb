## Basic Usage

The next five code cells show you how to include a graph-element, and add and remove vertices, while the Options section shows you what you can change about vertices and edges, either when you create them, or during an element's life time. 

```javascript
const { GraphElement } = await import("https://unpkg.com/@verticesandedges/graph-element/dist/index.js");

graph = document.createElement('graph-el');
graph.style.height = "500px";

return graph;
```

```javascript
a = graph.addVertex();
```

```javascript
e = graph.addEdge(a.id, a.id, { spline: true });
```

```javascript
graph.removeEdge(e);
```

```javascript
graph.removeVertex(a);
```

## Options

The next few cells show you how to customize your vertices and edges.

```javascript
a = graph.addVertex({
  size: 3.0,
  color: 'red', 
  texture: "https://picsum.photos/150/150",
  visible: true,
  selected: true,
  selectionColor: "black",
  label: "Hello, World!"
});

b = graph.addVertex({ color: "blue" });

e = graph.addEdge(a, b, {
  color: "orange",
  strength: 1.0,
  arrow: true,
  spline: false,
  width: 3.0,
  visible: true,
  label: "Greeting"
});
```
